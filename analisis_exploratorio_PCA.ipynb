<h4 style="color:#0F3D36; font-family:'Montserrat', sans-serif; font-size:20px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Metodología 1: One Hot Encoding + Standar Scaler + PCA
</h4>
<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Codificación de variables
</h4>

Ya que en la base de datos se tienen variables con tipos de datos numéricos y strings, es preciso codificar estas últimas para poder realizar los siguientes análisis. Se propone, para codificar la variable de $\texttt{nivel\_de\_digitalizacion}$, el método de one hot encoding; sin embargo, ya que este método transforma cada elemento en puntos ortonormales entre sí, de un espacio de dimensión $n$ donde $n$ es el número de elementos, cada uno de estos mantienen la misma norma respecto al origen, perdiendo información sobre la jerarquía de estos. 

```python
df['TraCosto'] = (df['TraCosto'] == 'VERDADERO').astype(int)

#One Hot Encoding para "nivel_digitalizacion"

columnas_a_codificar = ['nivel_digitalizacion']
df_exp1 = pd.get_dummies(df, columns=columnas_a_codificar, drop_first=False, dummy_na=False)

df_exp1.head()
```

Standar Scaler transoforma las variables que se le indique para que tengan media 0 y varianza 1

```python
columnas_base = [
    'Tiempo_en_minutos', 
    'N_FORMATOS_FINAL', 
    'CONTEO_NETO', 
    'TraCosto'
]

columnas_dig = [col for col in df_exp1.columns if 'nivel_digitalizacion' in col]


columnas_modelo = columnas_base + columnas_dig


scaler = StandardScaler()
datos_escalados = scaler.fit_transform(df_exp1[columnas_modelo])

df_escalado = pd.DataFrame(datos_escalados, columns=columnas_modelo)

df_escalado.head()
```

<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Principal Component Analysis
</h4>

```python
pca = PCA()


componentes = pca.fit_transform(df_escalado)

df_exp1['Indice_Complejidad'] = componentes[:, 0]

df_exp1.head()
```

```python
eigenvalores = pca.explained_variance_
varianza_explicada = pca.explained_variance_ratio_
varianza_acumulada = varianza_explicada.cumsum()

print("Componente | Eigenvalor | Varianza Explicada | Varianza Acumulada")
print("-" * 65)
for i in range(3):
    print(f"   PC{i+1}      | {eigenvalores[i]:>10.4f} | {varianza_explicada[i]:>17.4%} | {varianza_acumulada[i]:>17.4%}")
```

<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
H1: LA complejidad aumenta con más tiempo
</h4>

```python
slope, intercept, r_value, p_value, std_err = stats.linregress(
    df_exp1['Tiempo_en_minutos'], 
    df_exp1['Indice_Complejidad']
)

print("── H1: La complejidad aumenta con más tiempo ──")
print(f"  Pendiente:   {slope:.6f}")
print(f"  R²:          {r_value**2:.4f}")
print(f"  p-valor:     {p_value:.4e}")
print(f"  Conclusión:  {'✓ Significativa (p < 0.05)' if p_value < 0.05 else '✗ No significativa (p ≥ 0.05)'}")
```

<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
H2: LA complejidad aumenta con más requisitos / formatos
</h4>

```python
slope, intercept, r_value, p_value, std_err = stats.linregress(
    df_exp1['CONTEO_NETO'], 
    df_exp1['Indice_Complejidad']
)

print("── H2: La complejidad aumenta con más requisitos ──")
print(f"  Pendiente:   {slope:.6f}")
print(f"  R²:          {r_value**2:.4f}")
print(f"  p-valor:     {p_value:.4e}")
print(f"  Conclusión:  {'✓ Significativa (p < 0.05)' if p_value < 0.05 else '✗ No significativa (p ≥ 0.05)'}")
```

<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
H3: La complejidad disminuye con mayor digitalización
</h4>

```python
from sklearn.linear_model import LinearRegression
import numpy as np

columnas_dig = [col for col in df_exp1.columns if 'nivel_digitalizacion' in col]

X = df_exp1[columnas_dig].values
y = df_exp1['Indice_Complejidad'].values

from scipy import stats
X_con = np.column_stack([np.ones(len(X)), X])

coefs, residuals, rank, sv = np.linalg.lstsq(X_con, y, rcond=None)
y_pred = X_con @ coefs
residuals = y - y_pred
n, k = X_con.shape
var_res = np.sum(residuals**2) / (n - k)
cov_matrix = var_res * np.linalg.pinv(X_con.T @ X_con)
se = np.sqrt(np.diag(cov_matrix))
t_stats = coefs / se
p_values = 2 * stats.t.sf(np.abs(t_stats), df=n-k)
r2 = 1 - np.sum(residuals**2) / np.sum((y - np.mean(y))**2)

print("── H3: La complejidad disminuye con mayor digitalización ──")
print(f"\n  R²: {r2:.4f}")
print(f"\n  {'Variable':<45} {'Coef':>10} {'p-valor':>12}")
print("  " + "-"*70)
nombres = ['Intercepto'] + columnas_dig
for nombre, coef, pval in zip(nombres, coefs, p_values):
    sig = '*' if pval < 0.05 else ''
    print(f"  {nombre:<45} {coef:>10.4f} {pval:>12.4e} {sig}")
```

<h4 style="color:#0F3D36; font-family:'Montserrat', sans-serif; font-size:20px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Metodología 2: Ordinal Encoding + Robust Scaler + PCA
</h4>
<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Codificación de variables
</h4>

Como en la metodología anterior, al aplicar el One Hot Encoding en la variable de digitalización se perdía la jerarquía en los niveles, se propone utilizar un Ordinal Encoding tal que $\rm{Nivel}\;1 <\;....\;<\rm{Nivel}\;4.3$. Así mismo, para la estandarización de variables, debido a la presencia de outliers extremos encontrados previamente en algunas de las variables, se plantea el uso de RobustScaler. 

<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Codificación Ordinal
</h4>

A diferencia de la Metodología 1, donde se aplicó One Hot Encoding y se perdió la jerarquía entre niveles, aquí se utiliza **Ordinal Encoding** sobre `nivel_digitalizacion`, asignando valores enteros del 1 al 14 que preservan el orden natural de los niveles: a mayor valor, mayor digitalización. Del mismo modo, `TraCosto` se binariza (VERDADERO → 1, FALSO → 0).

```python
orden_dig = {

    'Nivel 1' : 1, 'Nivel 2': 2, 'Nivel 3.1' : 3, 'Nivel 3.2': 4, 'Nivel 3.3' : 5, 'Nivel 3.4' : 6, 'Nivel 3.5' : 7, 'Nivel 3.6' : 8, 'Nivel 3.7' : 9, 'Nivel 3.8' : 10, 'Nivel 3.9' : 11, 'Nivel 4.1' : 12, 'Nivel 4.2' : 13, 'Nivel 4.3' : 14  

}

df_exp2 = df.copy()
df_exp2['nivel_digitalizacion'] = df_exp2['nivel_digitalizacion'].map(orden_dig)

df_exp2.head().round(0)
```

<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Escalado con RobustScaler
</h4>

Dado que se detectaron outliers extremos en variables como `CONTEO_NETO` (2.41%) y `visitas_totales` (4.06%), se optó por **RobustScaler** en lugar de StandardScaler. Esta transformación centra los datos usando la mediana y los escala con el rango intercuartílico (IQR), haciéndola insensible a valores extremos:

$$x' = \frac{x - \text{mediana}}{IQR}$$

```python
Cols_M2 = [
    'Tiempo_en_minutos', 
    'N_FORMATOS_FINAL', 
    'CONTEO_NETO', 
    'TraCosto',
    'nivel_digitalizacion'
]

scaler_A = RobustScaler()
datos_esc_A = scaler_A.fit_transform(df_exp2[Cols_M2])
df_esc_A = pd.DataFrame(datos_esc_A, columns=Cols_M2, index=df_exp2.index)
```

<h4 style="color:#98989A; font-family:'Montserrat', sans-serif; font-size:16px; font-weight:600; margin-top:15px; margin-bottom:8px;">
Análisis de Componentes Principales
</h4>

Se aplica PCA sobre los datos escalados con RobustScaler. A diferencia de la Metodología 1, el escalado con RobustScaler no garantiza varianza unitaria por variable, por lo que el PCA opera sobre la **matriz de covarianza** y no sobre la de correlación. Esto produce eigenvalores en una escala distinta, no comparables directamente con los de la Metodología 1.

```python
pca_A = PCA()
componentes_A = pca_A.fit_transform(df_esc_A)
df_exp2['Indice_Esfuerzo_A'] = componentes_A[:, 0]
```

```python
eigenvalores = pca_A.explained_variance_
varianza_explicada = pca_A.explained_variance_ratio_
varianza_acumulada = varianza_explicada.cumsum()

print("Componente | Eigenvalor | Varianza Explicada | Varianza Acumulada")
print("-" * 65)
for i in range(3):
    print(f"   PC{i+1}      | {eigenvalores[i]:>10.4f} | {varianza_explicada[i]:>17.4%} | {varianza_acumulada[i]:>17.4%}")
```

```python
# ── Descomposición de componentes principales (Metodología 2) ──────────────
# pca_A.components_ → shape (n_PCs, n_variables)
# Cada fila es un PC, cada columna es la contribución (loading) de esa variable

Cols_M2 = [
    'Tiempo_en_minutos',
    'N_FORMATOS_FINAL',
    'CONTEO_NETO',
    'TraCosto',
    'nivel_digitalizacion'
]

n_pcs = len(Cols_M2)

df_loadings = pd.DataFrame(
    pca_A.components_,
    index   = [f'PC{i+1}' for i in range(n_pcs)],
    columns = Cols_M2
).round(4)

# Añadir columna de varianza explicada para contexto
df_loadings.insert(0, 'Varianza_explicada (%)',
                   (pca_A.explained_variance_ratio_ * 100).round(2))

print('Loadings por componente principal')
print('(qué tanto aporta cada variable a cada PC)\n')
display(df_loadings)

# ── Heatmap ─────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 4))

sns.heatmap(
    df_loadings[Cols_M2],
    annot      = True,
    fmt        = '.3f',
    cmap       = 'RdBu',
    center     = 0,
    linewidths = 0.5,
    ax         = ax,
    cbar_kws   = {'label': 'Loading'}
)

ax.set_title('Loadings — Contribución de cada variable a cada PC', fontsize=12, pad=12)
ax.set_xlabel('Variable original')
ax.set_ylabel('Componente principal')
plt.tight_layout()
plt.show()
```

El **PC1** explica el $53.61\%$ de la varianza con un eigenvalor de $10.48$, dominado por `nivel_digitalizacion` (loading = 0.994). El **PC2** explica el $33.55\%$ con un eigenvalor de $6.56$, dominado por `Tiempo_en_minutos` (loading = 0.994). Juntos acumulan el $87.15\%$ de la varianza.

Este patrón de loadings —donde cada PC captura casi exclusivamente una variable— es evidencia directa de que las variables **no están correlacionadas entre sí**: el PCA no logra combinarlas porque no hay estructura compartida que comprimir. Se necesitan los 5 componentes para representar los datos, lo que confirma la independencia estadística entre las variables del modelo.
